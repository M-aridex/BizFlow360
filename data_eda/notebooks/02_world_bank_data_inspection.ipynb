```python

print("=" * 70)
print("WORLD BANK DATA CLEANING - DUPLICATE REMOVAL")
print("=" * 70)

# Create a working copy
world_bank_clean_df = world_bank_df.copy()

# Check duplicates before removal
duplicates_before = world_bank_clean_df.duplicated().sum()

print("\nDuplicate rows before cleaning:", duplicates_before)

# Remove exact duplicate rows
world_bank_clean_df = world_bank_clean_df.drop_duplicates().reset_index(drop=True)

# Check duplicates after removal
duplicates_after = world_bank_clean_df.duplicated().sum()

print("Duplicate rows after cleaning:", duplicates_after)

print("\nOriginal shape:", world_bank_df.shape)
print("Cleaned shape :", world_bank_clean_df.shape)
print("Rows removed  :", world_bank_df.shape[0] - world_bank_clean_df.shape[0])

print("\n" + "=" * 70)
print("DUPLICATE REMOVAL COMPLETE")
print("=" * 70)
```

```python

import numpy as np
import pandas as pd

print("=" * 70)
print("WORLD BANK MISSING VALUE ANALYSIS")
print("=" * 70)

# Create a cleaning copy
world_bank_clean = world_bank_df.copy()

# ------------------------------------------------------------
# 1. Check for World Bank missing-value symbols
# ------------------------------------------------------------

missing_symbols = ["..", "...", "NA", "N/A", ""]

print("\n========== MISSING VALUE SYMBOL CHECK ==========")

for symbol in missing_symbols:
    count = (world_bank_clean == symbol).sum().sum()
    if count > 0:
        print(f"{repr(symbol)} values found: {count}")

# ------------------------------------------------------------
# 2. Convert World Bank missing symbols to NaN
# ------------------------------------------------------------

world_bank_clean = world_bank_clean.replace(
    ["..", "...", "NA", "N/A", ""],
    np.nan
)

print("\n✅ World Bank missing-value symbols converted to NaN.")

# ------------------------------------------------------------
# 3. Missing values by column
# ------------------------------------------------------------

missing_count = world_bank_clean.isna().sum()
missing_percentage = (
    world_bank_clean.isna().mean() * 100
).round(2)

missing_summary = pd.DataFrame({
    "missing_count": missing_count,
    "missing_percentage": missing_percentage
})

missing_summary = missing_summary.sort_values(
    "missing_count",
    ascending=False
)

print("\n========== MISSING VALUES BY COLUMN ==========")
print(missing_summary)

# ------------------------------------------------------------
# 4. Dataset shape
# ------------------------------------------------------------

print("\n========== DATASET STATUS ==========")
print("Dataset shape:", world_bank_clean.shape)
print("Total missing values:", world_bank_clean.isna().sum().sum())

print("\n" + "=" * 70)
print("MISSING VALUE ANALYSIS COMPLETE")
print("=" * 70)
```

```python

print("=" * 70)
print("WORLD BANK INDICATOR AVAILABILITY")
print("=" * 70)

# Count non-missing observations for every indicator
indicator_availability = (
    world_bank_clean
    .groupby(["Series Code", "Series Name"], dropna=False)
    .apply(
        lambda x: x.iloc[:, 4:].notna().sum().sum(),
        include_groups=False
    )
    .reset_index(name="available_observations")
)

# Total possible observations per indicator
number_of_years = 10
number_of_countries = world_bank_clean["Country Code"].nunique()

indicator_availability["possible_observations"] = (
    number_of_years * number_of_countries
)

indicator_availability["availability_percentage"] = (
    indicator_availability["available_observations"]
    / indicator_availability["possible_observations"]
    * 100
).round(2)

# Sort from most available to least available
indicator_availability = indicator_availability.sort_values(
    "availability_percentage",
    ascending=False
)

print("\n========== INDICATOR AVAILABILITY ==========")

print(
    indicator_availability.to_string(index=False)
)

print("\n========== SUMMARY ==========")

print(
    "Total indicators:",
    indicator_availability.shape[0]
)

print(
    "Indicators with >= 50% availability:",
    (
        indicator_availability["availability_percentage"] >= 50
    ).sum()
)

print(
    "Indicators with >= 25% availability:",
    (
        indicator_availability["availability_percentage"] >= 25
    ).sum()
)

print(
    "Indicators with >= 10% availability:",
    (
        indicator_availability["availability_percentage"] >= 10
    ).sum()
)

print("\n" + "=" * 70)
print("INDICATOR AVAILABILITY ANALYSIS COMPLETE")
print("=" * 70)
```

```python

print("=" * 70)
print("WORLD BANK SME-RELEVANT INDICATOR SELECTION")
print("=" * 70)

# Keywords related to SME financial health and business environment
relevant_keywords = [
    "finance",
    "investment",
    "loan",
    "collateral",
    "sales",
    "employment",
    "growth",
    "profit",
    "constraint",
    "corruption",
    "tax",
    "electricity",
    "informal",
    "ownership",
    "fixed assets",
    "working capital",
    "bank",
    "credit",
    "security"
]

# Combine indicator name and code for searching
indicator_text = (
    indicator_availability["Series Name"]
    .fillna("")
    .str.lower()
)

# Find potentially relevant indicators
relevant_mask = indicator_text.apply(
    lambda x: any(keyword in x for keyword in relevant_keywords)
)

world_bank_relevant_indicators = indicator_availability[
    relevant_mask
].copy()

print("\n========== RELEVANT INDICATORS ==========")

print(
    world_bank_relevant_indicators[
        [
            "Series Code",
            "Series Name",
            "available_observations",
            "availability_percentage"
        ]
    ].to_string(index=False)
)

print("\n========== SUMMARY ==========")

print(
    "Total indicators in World Bank dataset:",
    len(indicator_availability)
)

print(
    "Potentially relevant indicators:",
    len(world_bank_relevant_indicators)
)

print("\n" + "=" * 70)
print("INDICATOR SELECTION COMPLETE")
print("=" * 70)
```

```python

print("=" * 70)
print("CREATING WORLD BANK SELECTED INDICATOR DATASET")
print("=" * 70)

# Get the Series Codes selected in the previous step
selected_series_codes = (
    world_bank_relevant_indicators["Series Code"]
    .dropna()
    .unique()
    .tolist()
)

print("\nNumber of selected indicators:", len(selected_series_codes))

# Filter the cleaned World Bank dataset
world_bank_selected = world_bank_clean[
    world_bank_clean["Series Code"].isin(selected_series_codes)
].copy()

print("\n========== SELECTED DATASET ==========")

print("Shape:", world_bank_selected.shape)

print(
    "Number of countries:",
    world_bank_selected["Country Code"].nunique()
)

print(
    "Number of indicators:",
    world_bank_selected["Series Code"].nunique()
)

print(
    "Number of country names:",
    world_bank_selected["Country Name"].nunique()
)

# ------------------------------------------------------------
# Check observations by indicator
# ------------------------------------------------------------

selected_indicator_summary = (
    world_bank_selected
    .groupby(["Series Code", "Series Name"], dropna=False)
    .apply(
        lambda x: x.iloc[:, 4:].notna().sum().sum(),
        include_groups=False
    )
    .reset_index(name="available_observations")
)

selected_indicator_summary = (
    selected_indicator_summary
    .sort_values(
        "available_observations",
        ascending=False
    )
)

print("\n========== SELECTED INDICATOR COVERAGE ==========")

print(
    selected_indicator_summary.to_string(index=False)
)

print("\n" + "=" * 70)
print("SELECTED WORLD BANK DATASET CREATED")
print("=" * 70)
```

```python

print("=" * 70)
print("RESHAPING WORLD BANK DATA TO COUNTRY-YEAR FORMAT")
print("=" * 70)

# Make a copy
world_bank_long = world_bank_selected.copy()

# Year columns
year_columns = [
    "2011 [YR2011]",
    "2012 [YR2012]",
    "2013 [YR2013]",
    "2014 [YR2014]",
    "2015 [YR2015]",
    "2016 [YR2016]",
    "2017 [YR2017]",
    "2018 [YR2018]",
    "2019 [YR2019]",
    "2020 [YR2020]"
]

# ------------------------------------------------------------
# Convert wide format to long format
# ------------------------------------------------------------

world_bank_long = world_bank_long.melt(
    id_vars=[
        "Country Name",
        "Country Code",
        "Series Name",
        "Series Code"
    ],
    value_vars=year_columns,
    var_name="Year",
    value_name="Value"
)

# ------------------------------------------------------------
# Clean the Year column
# ------------------------------------------------------------

world_bank_long["Year"] = (
    world_bank_long["Year"]
    .str.extract(r"(\d{4})")[0]
    .astype(int)
)

# ------------------------------------------------------------
# Convert values to numeric
# ------------------------------------------------------------

world_bank_long["Value"] = pd.to_numeric(
    world_bank_long["Value"],
    errors="coerce"
)

print("\n========== LONG FORMAT ==========")

print("Shape:", world_bank_long.shape)

print(
    "Countries:",
    world_bank_long["Country Code"].nunique()
)

print(
    "Indicators:",
    world_bank_long["Series Code"].nunique()
)

print(
    "Years:",
    sorted(world_bank_long["Year"].unique())
)

print("\n========== FIRST 10 ROWS ==========")

print(
    world_bank_long.head(10).to_string(index=False)
)

print("\n========== MISSING VALUES ==========")

print(
    world_bank_long.isna().sum()
)

print("\n" + "=" * 70)
print("WORLD BANK RESHAPING COMPLETE")
print("=" * 70)
```

```python

print("=" * 70)
print("WORLD BANK COUNTRY-YEAR COVERAGE")
print("=" * 70)

# ------------------------------------------------------------
# 1. Count available indicator observations by country and year
# ------------------------------------------------------------

country_year_coverage = (
    world_bank_long
    .groupby(
        ["Country Code", "Country Name", "Year"]
    )["Value"]
    .count()
    .reset_index(name="available_indicators")
)

# Maximum possible indicators = 52
country_year_coverage["coverage_percentage"] = (
    country_year_coverage["available_indicators"]
    / 52
    * 100
).round(2)

# ------------------------------------------------------------
# 2. Display coverage summary
# ------------------------------------------------------------

print("\n========== COUNTRY-YEAR COVERAGE SUMMARY ==========")

print(
    country_year_coverage["available_indicators"]
    .describe()
)

print("\nCoverage percentage:")
print(
    country_year_coverage["coverage_percentage"]
    .describe()
)

# ------------------------------------------------------------
# 3. Best country-year observations
# ------------------------------------------------------------

print("\n========== TOP COUNTRY-YEAR OBSERVATIONS ==========")

print(
    country_year_coverage
    .sort_values(
        "available_indicators",
        ascending=False
    )
    .head(20)
    .to_string(index=False)
)

# ------------------------------------------------------------
# 4. Countries with the most available data
# ------------------------------------------------------------

country_coverage = (
    world_bank_long
    .groupby(
        ["Country Code", "Country Name"]
    )["Value"]
    .count()
    .reset_index(name="available_observations")
)

country_coverage["coverage_percentage"] = (
    country_coverage["available_observations"]
    / (52 * 10)
    * 100
).round(2)

print("\n========== COUNTRY COVERAGE ==========")

print(
    country_coverage
    .sort_values(
        "coverage_percentage",
        ascending=False
    )
    .head(20)
    .to_string(index=False)
)

print("\n" + "=" * 70)
print("COUNTRY-YEAR COVERAGE CHECK COMPLETE")
print("=" * 70)
```

```python

print("=" * 70)
print("WORLD BANK COUNTRY-YEAR COVERAGE FILTERING")
print("=" * 70)

# Check the coverage dataframe
print("\nOriginal country-year observations:", len(country_year_coverage))

# Coverage threshold
MIN_INDICATORS = 5

# Keep country-years with at least 5 available indicators
wb_usable_coverage = country_year_coverage[
    country_year_coverage["available_indicators"] >= MIN_INDICATORS
].copy()

print("\n========== COVERAGE FILTER ==========")

print("Minimum indicators required:", MIN_INDICATORS)

print("Usable country-year observations:",
      len(wb_usable_coverage))

print("Percentage retained:",
      round(
          len(wb_usable_coverage) /
          len(country_year_coverage) * 100,
          2
      ), "%")

print("\nAvailable indicators distribution:")

print(
    wb_usable_coverage["available_indicators"]
    .describe()
)

print("\n========== TOP COUNTRY-YEAR OBSERVATIONS ==========")

print(
    wb_usable_coverage[
        [
            "Country Name",
            "Country Code",
            "Year",
            "available_indicators",
            "coverage_percentage"
        ]
    ]
    .sort_values(
        "available_indicators",
        ascending=False
    )
    .head(20)
    .to_string(index=False)
)

print("\n" + "=" * 70)
print("COVERAGE FILTER COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# CREATE FINAL WORLD BANK FILTERED MODELING DATASET
# ======================================================================

print("=" * 70)
print("CREATING FINAL WORLD BANK MODELING DATASET")
print("=" * 70)

# Create a unique list of usable country-year combinations
usable_keys = wb_usable_coverage[
    ["Country Name", "Country Code", "Year"]
].drop_duplicates()

# Keep only observations belonging to usable country-years
world_bank_model_long = world_bank_long.merge(
    usable_keys,
    on=["Country Name", "Country Code", "Year"],
    how="inner"
)

print("\n========== FINAL FILTERED LONG DATA ==========")

print("Shape:", world_bank_model_long.shape)

print(
    "Country-year observations:",
    world_bank_model_long[
        ["Country Code", "Year"]
    ].drop_duplicates().shape[0]
)

print(
    "Countries:",
    world_bank_model_long["Country Code"].nunique()
)

print(
    "Indicators:",
    world_bank_model_long["Series Code"].nunique()
)

print(
    "Years:",
    sorted(world_bank_model_long["Year"].unique())
)

print("\n========== MISSING VALUES ==========")

print(
    world_bank_model_long.isna()
    .sum()
    .sort_values(ascending=False)
)

print("\n" + "=" * 70)
print("FINAL FILTERED WORLD BANK DATASET CREATED")
print("=" * 70)
```

```python

print("=" * 70)
print("RESHAPING FINAL WORLD BANK MODELING DATASET")
print("=" * 70)

# Pivot indicators into columns
world_bank_model_wide = world_bank_model_long.pivot_table(
    index=["Country Name", "Country Code", "Year"],
    columns="Series Code",
    values="Value",
    aggfunc="first"
).reset_index()

# Remove the columns-axis name created by pivot_table
world_bank_model_wide.columns.name = None

print("\n========== FINAL WIDE DATASET ==========")

print("Shape:", world_bank_model_wide.shape)

print(
    "Countries:",
    world_bank_model_wide["Country Code"].nunique()
)

print(
    "Country-year observations:",
    world_bank_model_wide[
        ["Country Code", "Year"]
    ].drop_duplicates().shape[0]
)

print(
    "Number of World Bank indicators:",
    len(world_bank_model_wide.columns) - 3
)

print("\n========== FIRST 5 ROWS ==========")

print(
    world_bank_model_wide.head().to_string(index=False)
)

print("\n========== MISSING VALUES ==========")

missing_wb_final = (
    world_bank_model_wide.isna()
    .sum()
    .sort_values(ascending=False)
)

print(missing_wb_final[missing_wb_final > 0])

print("\n========== DUPLICATE COUNTRY-YEARS ==========")

duplicate_country_years = (
    world_bank_model_wide
    .duplicated(
        subset=["Country Code", "Year"]
    )
    .sum()
)

print(
    "Duplicate country-year rows:",
    duplicate_country_years
)

print("\n" + "=" * 70)
print("WORLD BANK WIDE FORMAT COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# WORLD BANK MISSING VALUE IMPUTATION
# ======================================================================

print("=" * 70)
print("WORLD BANK MISSING VALUE IMPUTATION")
print("=" * 70)

# Create a copy
world_bank_imputed = world_bank_model_wide.copy()

# Identifier columns - DO NOT IMPUTE
identifier_columns = [
    "Country Name",
    "Country Code",
    "Year"
]

# World Bank indicator columns
indicator_columns = [
    col for col in world_bank_imputed.columns
    if col not in identifier_columns
]

print("\nNumber of indicator columns:", len(indicator_columns))

# Missing values BEFORE imputation
missing_before = (
    world_bank_imputed[indicator_columns]
    .isna()
    .sum()
    .sort_values(ascending=False)
)

missing_before = missing_before[missing_before > 0]

print("\n========== MISSING VALUES BEFORE IMPUTATION ==========")

print(missing_before)

print(
    "\nTotal missing values:",
    int(missing_before.sum())
)

# Store original missing locations
missing_mask = world_bank_imputed[indicator_columns].isna()

# Median imputation
for column in indicator_columns:

    median_value = world_bank_imputed[column].median()

    world_bank_imputed[column] = (
        world_bank_imputed[column]
        .fillna(median_value)
    )

# Missing values AFTER imputation
missing_after = (
    world_bank_imputed[indicator_columns]
    .isna()
    .sum()
    .sort_values(ascending=False)
)

missing_after = missing_after[missing_after > 0]

print("\n========== MISSING VALUES AFTER IMPUTATION ==========")

if len(missing_after) == 0:
    print("✅ No missing values remain in World Bank indicators.")
else:
    print(missing_after)

print(
    "\nTotal missing values after imputation:",
    int(
        world_bank_imputed[indicator_columns]
        .isna()
        .sum()
        .sum()
    )
)

print("\n========== FINAL DATASET ==========")

print("Shape:", world_bank_imputed.shape)

print(
    "Countries:",
    world_bank_imputed["Country Code"].nunique()
)

print(
    "Country-year observations:",
    world_bank_imputed[
        ["Country Code", "Year"]
    ].drop_duplicates().shape[0]
)

print(
    "Indicators:",
    len(indicator_columns)
)

print("\n" + "=" * 70)
print("WORLD BANK IMPUTATION COMPLETE")
print("=" * 70)
```
